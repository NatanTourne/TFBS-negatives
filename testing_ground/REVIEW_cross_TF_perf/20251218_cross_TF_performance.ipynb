```python
import os
import pandas as pd
# List all files in the directory
df_folder = '/data/home/natant/Negatives/Runs/Review_sane_model/Cross_TF'
df_files = [f for f in os.listdir(df_folder) if f.endswith('.csv')]

# Load each CSV as a DataFrame and store in a dictionary, using the first column as index
cross_TF_dfs = {}
for fname in df_files:
    key = fname.replace('.csv', '')
    cross_TF_dfs[key] = pd.read_csv(os.path.join(df_folder, fname), index_col=0)

# Show loaded keys
print("Loaded DataFrames:", list(cross_TF_dfs.keys()))
```

```python
import numpy as np

# Define the different types of negative samples used in the analysis
negative_types = ["dinucl_sampled", "shuffled", "dinucl_shuffled", "celltype"]
# Define the metrics to analyze
metrics = ["AUROC", "AUROC_HQ"]
# Dictionary to store averaged dataframes across folds
averaged_dfs = {}

# Iterate through each negative type and metric combination
for neg in negative_types:
    for metric in metrics:
        # List to store dataframes from different CV folds
        dfs = []
        # Collect data from 6 cross-validation folds (0-5)
        for fold in range(6):
            # Construct the key to access the appropriate dataframe
            key = f"A549_{neg}_CV-{fold}_{metric}"
            # Check if the key exists in the loaded dataframes
            if key in cross_TF_dfs:
                dfs.append(cross_TF_dfs[key])
        
        print([df.shape for df in dfs])  # Print shapes of collected dataframes for debugging
        # If we found any dataframes for this combination
        if len(dfs) > 0:
            # Create a union of all indices and columns to ensure all dataframes have same dimensions
            all_idx = sorted(set().union(*[df.index for df in dfs]))
            all_cols = sorted(set().union(*[df.columns for df in dfs]))
            
            # Align all dataframes to have the same rows and columns
            aligned = [df.reindex(index=all_idx, columns=all_cols) for df in dfs]
            
            # Stack the aligned dataframes to compute the mean
            stacked = np.stack([df.values for df in aligned])
            
            # Calculate the mean across folds, ignoring NaN values
            avg_matrix = np.nanmean(stacked, axis=0)
            
            # Create a new dataframe with the averaged values
            avg_df = pd.DataFrame(avg_matrix, index=all_idx, columns=all_cols)
            
            # Store the averaged dataframe with a descriptive key
            averaged_dfs[f"{neg}_{metric}"] = avg_df
        else:
            # Warning if no data was found for a particular combination
            print(f"Warning: No folds found for {neg} {metric}")

# Return the dictionary of averaged dataframes
averaged_dfs
```

```python
# Create mapping dictionaries to handle underscores in names
neg_mode_mapping = {
    'dinucl_sampled': 'dinucl-sampled',
    'dinucl_shuffled': 'dinucl-shuffled',
    'shuffled': 'shuffled',
    'celltype': 'celltype'
}

metric_mapping = {
    'AUROC': 'AUROC',
    'AUROC_HQ': 'AUROC-HQ'
}

# Create reverse mappings for converting back
neg_mode_reverse = {v: k for k, v in neg_mode_mapping.items()}
metric_reverse = {v: k for k, v in metric_mapping.items()}

# Create a comprehensive dataframe with all cross-validation results
all_results = []

# Iterate through each dataframe in cross_TF_dfs
for key, df in cross_TF_dfs.items():
    # Replace underscores in the key for proper parsing
    modified_key = key
    for original, replacement in neg_mode_mapping.items():
        modified_key = modified_key.replace(original, replacement)
    for original, replacement in metric_mapping.items():
        modified_key = modified_key.replace(original, replacement)
    
    # Parse the modified key to extract metadata
    parts = modified_key.split('_')
    celltype = parts[0]  # A549
    neg_mode = neg_mode_reverse[parts[1]]  # Convert back to original
    cv_split = parts[2]  # e.g., 'CV-0', 'CV-1', etc.
    metric = metric_reverse[parts[3]]  # Convert back to original
    
    # Convert the cross-TF matrix to long format
    for train_tf in df.index:
        for test_tf in df.columns:
            score = df.loc[train_tf, test_tf]
            
            all_results.append({
                'celltype': celltype,
                'neg_mode': neg_mode,
                'cv_split': cv_split,
                'metric': metric,
                'train_tf': train_tf,
                'test_tf': test_tf,
                'score': score
            })

# Create the comprehensive dataframe
comprehensive_df = pd.DataFrame(all_results)

# Pivot to have AUROC and AUROC_HQ as separate columns
comprehensive_df = comprehensive_df.pivot_table(
    index=['celltype', 'neg_mode', 'cv_split', 'train_tf', 'test_tf'],
    columns='metric',
    values='score'
).reset_index()

# Flatten column names
comprehensive_df.columns.name = None

comprehensive_df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
def plot_side_by_side_heatmaps(df1, df2, title1=None, title2=None, figsize=(18, 8), 
                              use_global_scale=True, vmin=None, vmax=None):
    """
    Create two side-by-side heatmaps for comparing dataframes.
    
    Parameters:
    df1, df2: pandas DataFrames to plot
    title1, title2: titles for each heatmap (optional)
    figsize: figure size tuple
    use_global_scale: if True, use same color scale for both heatmaps
    vmin, vmax: custom min/max values for color scale (overrides use_global_scale)
    """
    # Determine TF order based on first dataframe's diagonal mean
    TF_order = df1.mean(axis=1).sort_values(ascending=False).index.to_list()
    
    # Reorder both dataframes
    df1_ordered = df1.loc[TF_order, TF_order]
    df2_ordered = df2.loc[TF_order, TF_order]
    
    # Set color scale
    if vmin is None or vmax is None:
        if use_global_scale:
            vmin = min(df1_ordered.min().min(), df2_ordered.min().min())
            vmax = max(df1_ordered.max().max(), df2_ordered.max().max())
        else:
            vmin, vmax = None, None
    
    # Create figure
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    
    # Create short labels
    short_labels_ordered = [label.split("_")[0] for label in TF_order]
    
    # Plot first heatmap
    sns.heatmap(df1_ordered, annot=True, cmap="RdYlBu", fmt=".2f",
                xticklabels=short_labels_ordered, yticklabels=short_labels_ordered, 
                ax=axes[0], vmin=vmin, vmax=vmax)
    axes[0].xaxis.set_ticks_position('top')
    axes[0].xaxis.set_label_position('top')
    if title1:
        axes[0].set_title(title1, pad=40)
    axes[0].set_ylabel("TF the model was trained on")
    axes[0].set_xlabel("Tested TF")
    axes[0].tick_params(axis='x', rotation=45, labelrotation=45)
    
    # Plot second heatmap
    sns.heatmap(df2_ordered, annot=True, cmap="RdYlBu", fmt=".2f",
                xticklabels=short_labels_ordered, yticklabels=short_labels_ordered, 
                ax=axes[1], vmin=vmin, vmax=vmax)
    axes[1].xaxis.set_ticks_position('top')
    axes[1].xaxis.set_label_position('top')
    if title2:
        axes[1].set_title(title2, pad=40)
    axes[1].set_ylabel("TF the model was trained on")
    axes[1].set_xlabel("Tested TF")
    axes[1].tick_params(axis='x', rotation=45, labelrotation=45)
    
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import pandas as pd

neg_mode_labels = {
    'dinucl_sampled': 'G-S',
    'dinucl_shuffled': 'DN-SH',
    'neighbors': 'NB-S',
    'shuffled': 'SH',
    'motif': "PWM",
    'celltype': 'CL',
    'HQ training': 'HQ'
}

# Plot order
hue_order = ['False_AUROC', 'True_AUROC', 'False_AUROC_HQ', 'True_AUROC_HQ']

# Color mapping
palette_map = {
    'False_AUROC': orange,
    'True_AUROC': orange_dark,
    'False_AUROC_HQ': green,
    'True_AUROC_HQ': green_dark
}

# Cross-TF boxes to hatch
hatched_hues = {'True_AUROC', 'True_AUROC_HQ'}

neg_modes = list(comprehensive_melted['neg_mode'].unique())
n_neg = len(neg_modes)
n_hue = len(hue_order)
width = 0.18
center_positions = np.arange(n_neg)

plt.figure(figsize=(6,5))

for i, neg in enumerate(neg_modes):
    for h_idx, hue in enumerate(hue_order):
        data = comprehensive_melted.loc[
            (comprehensive_melted['neg_mode'] == neg) &
            (comprehensive_melted['cross_TF_metric'] == hue), 'performance'
        ].dropna().values
        if len(data) == 0:
            continue
        pos = i + (h_idx - (n_hue-1)/2) * width
        bp = plt.boxplot(
            data,
            positions=[pos],
            widths=width*0.9,
            patch_artist=True,
            manage_ticks=False,
            medianprops=dict(color="black", linewidth=1.2)
        )
        # set color and hatch
        box_color = palette_map.get(hue, 'lightgray')
        for box in bp['boxes']:
            box.set_facecolor(box_color)
            box.set_edgecolor('black')
            if hue in hatched_hues:
                box.set_hatch('///')

# Rename neg_modes
renamed_neg_modes = [neg_mode_labels.get(neg, neg) for neg in neg_modes]

plt.xticks(center_positions, renamed_neg_modes)
plt.xlabel('Negative Sampling Mode')
plt.ylabel('Performance')
plt.ylim(0,1)  # y-axis from 0 to 1

# Custom legend inside bottom-right
legend_labels = ['Same TF (AUROC)', 'Cross TF (AUROC)',
                 'Same TF (AUROC-HQ)', 'Cross TF (AUROC-HQ)']
legend_handles = []
for hue, lab in zip(hue_order, legend_labels):
    col = palette_map[hue]
    hatch = '///' if hue in hatched_hues else ''
    legend_handles.append(mpatches.Patch(facecolor=col, edgecolor='black', hatch=hatch, label=lab))

plt.legend(handles=legend_handles, loc='lower right', title='Cross-TF & Metric')

plt.axhline(y=0.5, color='red', linestyle='-', linewidth=1)
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.savefig('/data/home/natant/Negatives/figures/performance_boxplot.eps', bbox_inches='tight')
plt.show()

```
