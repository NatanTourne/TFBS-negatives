```python
import os

base_dir = "/data/home/natant/Negatives/Runs/Review_sane_model/other_cell_lines"

try:
    files_and_dirs = os.listdir(base_dir)
    print(f"Contents of {base_dir}:")
    for name in files_and_dirs:
        print(name)
except FileNotFoundError:
    print(f"Path not found: {base_dir}")
except PermissionError:
    print(f"Permission denied: {base_dir}")
```

```python
import pandas as pd

rows = []
for fname in files_and_dirs:
    # remove extension
    base = fname.rsplit(".ckpt", 1)[0]
    parts = base.split("_")

    row = {"filename": fname}

    # first 6 parts: CT, TF, NEG, CV, LR, Date
    for part in parts:
        if "-" not in part:
            continue
        key, value = part.split("-", 1)

        if key == "CT":
            row["CT"] = value
        elif key == "TF":
            row["TF"] = value
        elif key == "NEG":
            row["NEG"] = value
        elif key == "CV":
            row["CV"] = int(value)
        elif key == "LR":
            row["LR"] = float(value)
        elif key == "Date":
            row["Date"] = value
        elif key == "epoch":
            row["Epoch"] = int(value)
        elif key == "val_loss":
            row["val_loss"] = float(value)
        elif key == "AUROC":
            row["AUROC"] = float(value)
        # ignore Time and anything else for now

    rows.append(row)

df = pd.DataFrame(rows)
df
```

```python
# Count how many entries there are for each TF-NEG combination
tf_neg_counts = HepG2.groupby("TF_NEG").size()

# Select combinations whose count is not 6
tf_neg_not_six = tf_neg_counts[tf_neg_counts != 6]

tf_neg_not_six
```

```python
# Unique TFs observed in HepG2
hep_tfs = HepG2["TF"].unique()

# All NEG types in the whole dataset
neg_types = df["NEG"].dropna().unique()

# Expected TF-NEG combos: all (HepG2 TF) x (NEG type)
expected_hep_tf_neg = {f"{tf}_{neg}" for tf in hep_tfs for neg in neg_types}

# What we actually have for HepG2
hep_tf_neg = set(HepG2["TF_NEG"].dropna())

# Combos that are missing HepG2 runs (i.e. never had any HepG2 CV folds)
no_cv_tf_neg = sorted(expected_hep_tf_neg - hep_tf_neg)

# Combos that have some folds but not 6 (from tf_neg_not_six)
some_but_not_six_tf_neg = sorted(tf_neg_not_six.index.tolist())

# All problematic combos: either zero folds or not exactly six
problem_tf_neg = sorted(set(no_cv_tf_neg) | set(some_but_not_six_tf_neg))

problem_tf_neg
```
